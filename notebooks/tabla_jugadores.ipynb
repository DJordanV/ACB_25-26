```python
df = pd.read_csv(
    DATA_PATH/'Jugadores.csv',
    dtype={
        "Equipo": "object",
        "Dorsal": "object",
        "Nombre": "object",
        "Apellido": "object",
        "Posición": "object",
        "Altura": "object",
        "Fecha_Nacim": "object",
        "Pais_Nacim": "object",
        "Comunidad": "object",
        "Nacionalidad": "object",
        "Licencia": "object"
    })
```

```python
jugadores = []

while True:

    equipo = 'JOV'
    # equipo = input('Equipo: ').upper()
    dorsal = input('Dorsal: ')
    nombre = input('Nombre (sin apellido): ').title()
    apellido = input('Apellido: ').title()
    posicion = input('Posición: ')
    altura = input('Altura en cm: ')

    dia_nacim = input('Día de nacimiento (DD): ')
    mes_nacim = input('Mes de nacimiento (MM): ')
    año_nacim = input('Año de nacimiento (AAAA): ')
    fecha_nacim = dia_nacim + '/' + mes_nacim + '/' + año_nacim
    
    pais_nacim = input('País de Nacimiento: ').upper()

    if pais_nacim in ['ESP', 'ESPAÑA']:
        comunidad = input('Comunidad Autónoma: ').upper()
    else:
        comunidad = ''

    nacionalidad = input('Nacionalidad: ').upper()
    licencia = input('Licencia: ').upper()

    nueva_fila = {
        'Equipo': equipo,
        'Dorsal': dorsal,
        'Nombre': nombre,
        'Apellido': apellido,
        'Posición': posicion,
        'Altura': altura,
        'Fecha_Nacim': fecha_nacim,
        'Pais_Nacim': pais_nacim,
        'Comunidad': comunidad,
        'Nacionalidad': nacionalidad,
        'Licencia': licencia
    }

    jugadores.append(nueva_fila)

    continuar = input('Añadir otro jugador? (S/N): ').upper()
    if continuar != 'S':
        break

df = pd.concat(
    [df, pd.DataFrame(jugadores)],
    ignore_index=True
)
```

```python
iso_paises = {
    'ESPAÑA': 'ESP',
    'ARGENTINA': 'ARG',
    'ITALIA': 'ITA',
    'CHEQUIA': 'CZE',
    'SENEGAL': 'SEN',
    'FRANCIA': 'FRA',
    'GEORGIA': 'GEO',
    'MALI': 'MLI',
    'ESTONIA': 'EST',
    'PAISES BAJOS': 'NLD',
    'SERBIA': 'SRB',
    'CROACIA': 'HRV',
    'HUNGRIA': 'HUN',
    'MONTENEGRO': 'MNE',
    'REINO UNIDO': 'GBR',
    'LITUANIA': 'LTU',
    'SUECIA': 'SWE'
}
df.replace(iso_paises, inplace=True)
```

```python
df['Foto'] = 'https://github.com/DJordanV/ACB_25-26/blob/develop/img/clubes/' + df['Equipo'] + '/cuerpo/' + df['Dorsal'] + '.webp?raw=true'
df['Bandera_Pais'] = 'https://github.com/DJordanV/ACB_25-26/blob/develop/img/banderas/' + df['Pais_Nacim'] + '.png?raw=true'
df['Bandera_Comunidad'] = np.where(df['Comunidad'] == '', '', 'https://github.com/DJordanV/ACB_25-26/blob/develop/img/banderas/ESP/' + df['Comunidad'] + '.png?raw=true')
```

```python
columnas = ["Pais_Nacim", "Comunidad", "Nacionalidad"]

for col in columnas:
    mask = df[col].notna() & (df[col].str.len() > 3)

    if mask.any():
        warnings.warn(
            f"La columna '{col}' contiene valores con más de 3 caracteres: "
            f"{df.loc[mask, col].unique().tolist()}",
            UserWarning
        )
```
